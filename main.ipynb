```python
# 기본지도
m = folium.Map(
    location = [33.364969, 126.529117],
    tiles = 'OpenStreetMap',
    zoom_start = 10
)

tooltip = "Click!!"

# 오름 정보 추가
for i in range(oreum.shape[0]):
    o_name = oreum.iloc[i]['오름명']
    lat = oreum.iloc[i]['위도']
    long = oreum.iloc[i]['경도']

    folium.Marker(
        [lat, long],
        popup = f"<strong>{o_name}</strong>",
        tooltip = tooltip
    ).add_to(m)

m
```

```python
# 오름 상세정보 추가하기

# 기본지도
m = folium.Map(
    location = [33.364969, 126.529117],
    tiles = 'OpenStreetMap',
    zoom_start = 10
)

tooltip = "Click!!"

# 오름 정보 추가
for i in range(oreum.shape[0]):
    folium.Marker(
        [oreum.iloc[i]['위도'], oreum.iloc[i]['경도']],
        popup = f"<strong>{oreum.iloc[i]['오름명']}</strong>",
        tooltip = tooltip
    ).add_to(m)

m
```

```python
# 오름 상세정보 추가하기

# 기본지도
m = folium.Map(
    location = [33.364969, 126.529117],
    tiles = 'OpenStreetMap',
    zoom_start = 10
)

tooltip = "Click!!"

# 오름 정보 추가
for i in range(oreum.shape[0]):
    folium.Marker(
        [oreum.iloc[i]['위도'], oreum.iloc[i]['경도']],
        popup = f"<div style='width:100px'><strong>{oreum.iloc[i]['오름명']}</strong><br>\
            주차장: {oreum.iloc[i]['주차장']}<br>\
            화장실: {oreum.iloc[i]['화장실']}<br>\
            </div>",
        tooltip = tooltip
    ).add_to(m)

m
```
